# Protein report template
Set `GENE` and run all cells. Needs the study's derived tables (`data/derived`). Optional: `ORTHOLOGUES = True` (network the first time).

```
import sys; sys.path.insert(0, '../../src')
import pandas as pd
from abhd.atlas import Atlas
from abhd.protein_report import build_report
from abhd.compare import compare, to_markdown
A = Atlas()
GENE = 'ABHD2'
OTHER = 'MGLL'
ORTHOLOGUES = False
```

## Facts

```
info = A.info(GENE)
pd.Series({k: v for k, v in info.items() if not isinstance(v, (list, dict))})
```

## Neighbours (nearest first); identity is a global alignment on the trimmed domains

```
for view in ['structure', 'sequence', 'funfam', 'elbow_window']:
    print(view); display(A.neighbours(GENE, view, k=8, cross_family=True)[['gene', 'distance', 'identity', 'abhd_named', 'ec3']])
```

## Function transfer (tiers: <=30% no gain; 30-40% unresolved; >=40% weak gain)

```
[A.predict_ec(GENE, v) for v in ('structure', 'sequence', 'funfam')]
```

## Full HTML report (and 3D view files)

```
res = build_report(A, GENE, '../../out/protein_reports', orthologues=ORTHOLOGUES, structure=True)
res['path']
```

## Compare with another protein

```
from IPython.display import Markdown
Markdown(to_markdown(compare(A, GENE, OTHER)))
```
